Task 1: Line Plot

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 170, 160, 200, 220]

plt.plot(months, sales, marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid(True)

plt.show()
```

Task 2: Multiple Line Plot

```python
import matplotlib.pyplot as plt

subjects = ["Math", "Physics", "Chemistry", "Biology", "English", "ICT"]
student_A = [85, 78, 90, 88, 76, 92]
student_B = [80, 82, 87, 85, 79, 89]

plt.plot(subjects, student_A, marker='o', linestyle='-', label='Student A')
plt.plot(subjects, student_B, marker='s', linestyle='--', label='Student B')

plt.title("Student Performance Comparison")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.legend()
plt.grid(True)

plt.show()
```

Task 3: Bar Chart

```python
import matplotlib.pyplot as plt

departments = ["CSE", "EEE", "BBA", "Civil", "English"]
students = [250, 180, 300, 120, 150]

bars = plt.bar(departments, students,
               color=['blue', 'green', 'orange', 'red', 'purple'])

for bar in bars:
    y = bar.get_height()
    plt.text(bar.get_x()+bar.get_width()/2,
             y+5,
             y,
             ha='center')

plt.title("Department-wise Student Enrollment")
plt.xlabel("Departments")
plt.ylabel("Number of Students")

plt.show()
```

Task 4: Histogram

```python
import matplotlib.pyplot as plt

marks = [65, 70, 75, 80, 85, 90, 72, 68, 95, 88,
         77, 83, 91, 73, 69, 81, 86, 92, 78, 74]

plt.figure(figsize=(8,5))

plt.hist(marks, bins=6, edgecolor='black')

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

Task 5: Pie Chart

```python
import matplotlib.pyplot as plt

activities = ["Study", "Sleep", "Exercise", "Entertainment", "Others"]
hours = [6, 8, 1, 4, 5]

# Highlight the largest segment (Sleep)
explode = [0, 0.1, 0, 0, 0]

plt.figure(figsize=(7,7))

plt.pie(hours,
        labels=activities,
        autopct='%1.1f%%',
        explode=explode,
        startangle=90)

plt.title("Daily Activity Distribution")

plt.show()
```

Task 6: Student Performance Dashboard (5 Marks)

```python
import matplotlib.pyplot as plt

# Data
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
attendance = [85, 88, 90, 87, 92, 95]

subjects = ["Math", "Physics", "Chemistry", "English", "ICT"]
marks = [88, 75, 82, 90, 95]

activities = ["Study", "Sleep", "Exercise", "Entertainment", "Others"]
hours = [6, 8, 1, 3, 6]

scores = [65, 70, 75, 80, 85, 90, 72, 68, 95, 88,
          77, 83, 91, 73, 69, 81, 86, 92, 78, 74]

# largest segment highlight
explode = [0, 0.1, 0, 0, 0]

# Dashboard
fig, ax = plt.subplots(2, 2, figsize=(12, 8))

#1. Line Plot: Attendance
ax[0, 0].plot(months, attendance, marker='o', color='blue')
ax[0, 0].set_title("Monthly Attendance")
ax[0, 0].set_xlabel("Months")
ax[0, 0].set_ylabel("Attendance (%)")
ax[0, 0].grid(True)

# 2. Bar Chart: Subject Marks
bars = ax[0, 1].bar(subjects, marks, color='orange')

# show values on top
for bar in bars:
    height = bar.get_height()
    ax[0, 1].text(bar.get_x() + bar.get_width()/2,
                  height + 1,
                  str(height),
                  ha='center')

ax[0, 1].set_title("Subject-wise Marks")
ax[0, 1].set_xlabel("Subjects")
ax[0, 1].set_ylabel("Marks")

# 3. Pie Chart: Daily Activities
ax[1, 0].pie(hours,
             labels=activities,
             autopct='%1.1f%%',
             explode=explode,
             startangle=90)

ax[1, 0].set_title("Daily Time Allocation")

# 4. Histogram: Exam Scores
ax[1, 1].hist(scores, bins=6, color='green', edgecolor='black')

ax[1, 1].set_title("Exam Score Distribution")
ax[1, 1].set_xlabel("Scores")
ax[1, 1].set_ylabel("Frequency")

# Layout
plt.tight_layout()

# Save figure
plt.savefig("student_dashboard.png", dpi=300)

# Show dashboard
plt.show()
```
