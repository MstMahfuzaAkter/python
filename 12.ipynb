```python
# Employee dictionary
employees = {
    "Ali": 80000,
    "Asif": 70000,
    "Karim": 65000
}

# Add a new employee
employees["Ratan"] = 75000

# Update salary of Ali
employees["Ali"] = 85000

# Remove employee Asif
employees.pop("Asif")

# Display all employees and salaries
print("Employee Salaries:")
for name, salary in employees.items():
    print(name, ":", salary)
```

```python
class Sitar:
    def play(self):
        print("Strumming the sitar")

class Tabla:
    def play(self):
        print("Playing the tabla")

class Dhol:
    def play(self):
        print("Beating the dhol")

# Create objects
s1 = Sitar()
t1 = Tabla()
d1 = Dhol()

# Store objects in a list
instruments = [s1, t1, d1]

# Demonstrate polymorphism
for instrument in instruments:
    instrument.play()
```

```python
def format_and_filter_names(names):
    result = []

    for name in names:

        # Remove names containing "masud"
        if "masud" in name.lower():
            continue

        # Convert to Proper Case
        parts = name.lower().split()
        parts = [p.capitalize() for p in parts]

        # Format: Last Middle First
        if len(parts) == 3:
            formatted = parts[2] + " " + parts[1] + " " + parts[0]
        elif len(parts) == 2:
            formatted = parts[1] + " " + parts[0]
        else:
            formatted = parts[0]

        result.append(formatted)

    return result


# Example
names = [
    "md ali hasan",
    "masud rana",
    "john smith",
    "rahim uddin karim"
]

print(format_and_filter_names(names))
```

```python
import matplotlib.pyplot as plt

semesters = ['Sem 1', 'Sem 2', 'Sem 3', 'Sem 4', 'Sem 5', 'Sem 6']
total_marks = [550, 600, 350, 450, 360, 370]

plt.plot(semesters, total_marks,
         marker='o', linewidth=2)

plt.title("Student Performance Over Semesters")
plt.xlabel("Semesters")
plt.ylabel("Total Marks")
plt.grid(True)

plt.show()
```
