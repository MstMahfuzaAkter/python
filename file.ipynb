```python
from functools import reduce
with open("./sample_data/sample.txt","r") as file:
  strings=file.readlines()
#total line
  total_lines=len(strings)
#each line  a word
  number_of_words=list(map(lambda x:len(x.split()),strings))
  #total number of words
  total_number_of_words=reduce(lambda x,y:x+y,number_of_words)

  print(strings)
  print(total_lines)
  print(number_of_words)
  print(total_number_of_words)


#number of characher
  #cleaning process

  #1.new line delete
  strings=list(map(str.strip,strings))
  print(strings)
#2.remove space

  strings=list(map(lambda x: x.replace(" ",""),strings))
  print(strings)

  for i in strings:
    print(i)

  #koita character
  number_of_char =list(map(lambda x:len(x),strings))
  print(number_of_char)

  #total number of character

  total_char=reduce(lambda x,y:x+y,number_of_char)
  print(total_char)



  with open("./sample_data/counter_of_string.txt","w") as file:
    file.write(f"total line:{total_lines}\n total_number_of_words:{total_number_of_words}\n total number of char:{total_char}")

```
