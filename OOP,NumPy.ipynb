Class and Objects

```python
# class attribute
class Phone:
    category = "Electronics"

    # constructor
    def __init__(self, model, battery, camera, battery_percentage=100):
        self.model = model
        self.battery = battery
        self.camera = camera
        self.battery_percentage = battery_percentage

    # method
    def charge(self, hour):
        print(f"Charge is complete: {hour} hour(s)")

    def capture(self, total_photo):
        if self.battery_percentage <= 0:
            print("No charge")
        else:
            self.battery_percentage -= total_photo
            print(f"Photo Capture in {self.model}")


class Cooling_Mechanism:
    def __init__(self, cooling_method):
        self.cooling_method = cooling_method

    def cooling_on(self):
        print(f"The system is being cooled by {self.cooling_method}")
```

```python
#creating object

#apple=Phone()
apple.category="Super Electronics"
#print(type(apple))
#blueberry=Phone()
#motorola=Phone()


print(apple.category)
#print(blueberry.category)
#print(motorola.category)

apple=Phone("Iphone17",3000,40)
blueberry=Phone("B-17",4000,30)
motorola=Phone("M-17",3500,35)

print(apple.camera)

print(motorola.battery_percentage)

motorola.capture(10)
print(motorola.battery_percentage)

```

OOP Inheritance

```python
class Smartphone(Phone):
    def __init__(self, model, battery, camera, processor):
        super().__init__(model, battery, camera)
        self.processor = processor

    def charge(self, hour):
        print("Fast charging in process")
        super().charge(hour)
```

```python
pro=Smartphone("X",5000,100,"SnapDragon")

print(pro.model)
pro.charge(1)

```

Multiple Inheritance

```python
class SmartPhn_Cooling(Smartphone,Cooling_Mechanism):
  def __init__(self, model, battery, camera, processor,cooling_method):
     Smartphone.__init__(self,model, battery, camera, processor)
     Cooling_Mechanism.__init__(self,cooling_method)
```

```python
pro_cooling=SmartPhn_Cooling("Y",5000,30,"SnapDragon","Nitrogen")
print(pro_cooling.processor)

print(pro_cooling.cooling_on())
pro_cooling.charge(1)
```

OOP Polymorphism-->overriding

```python
class Camera:
  def __init__(self,name):
      self.name=name
  def capture(self):
    print("a photo is capture")
```

```python
class Smart_Phone(Camera):
  def __init__(self, name,resolution):
      super().__init__(name)
      self.resolution=resolution

  #method override
  def capture(self):
     print("Photo is capture by a phone")
class DSLR(Camera):
  def __init__(self,name,resolution):
    super().__init__(name)
    self.resolution=resolution

  def capture(self):
     print("Photo is capture by a dslr")

class Drone(Camera):
  def __init__(self,name,resolution):
    super().__init__(name)
    self.resolution=resolution

  def capture(self):
     print("Photo is capture by a drone")

```

```python
phone=Smart_Phone("Phpne",60)
dslr=DSLR("DSLR",200)
drone=Drone("Drone",400)

phone.capture()
dslr.capture()
drone.capture()
```

 Abstraction--->complex method ke hide kore operate korer jonno funtion/method provide korbo

```python
from abc import ABC, abstractmethod

class Telephone(ABC):
    @abstractmethod
    def make_call(self):
        pass

    @abstractmethod
    def capture_photo(self):
        pass


class Sphone(Telephone):
    def make_call(self):
        print("Making a call from Sphone")

    def capture_photo(self):
        print("Sphone capturing photo")


class Iphone(Telephone):
    def make_call(self):
        print("Making a call from Iphone")

    def capture_photo(self):
        print("Iphone capturing photo")
```

```python
ip=Iphone()
ip.make_call()
```
