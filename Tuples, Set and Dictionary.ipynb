```python
print(students.get("height"))
```

```python
co_ordinates=[(10,10.5),(20.5,192),(101,102)]
location=["Dhaka","Shariatpur","Sylhet"]

exact_location=[co_or:loc for co_or,
loc in co_ordinates,location]
print(exact_location)
```

```python
co_ordinates=[(10,10.5),(20.5,192),(101,102)]
location=["Dhaka","Shariatpur","Sylhet"]

exact_location={co_or:loc for co_or,
                loc in zip(co_ordinates,
                           location)}
print(exact_location)
```
