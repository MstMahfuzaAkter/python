```python
def square_addition(*args):
    sum=0 
    for i in args:
        i=i**2
        sum+=i
    return sum          
ans=square_addition(10,20,30,40)     
print(ans)  
```

```python
lst=[x for x in range(500)]#dataset
def data_loader(chunk_size,lst):
    for i in range(0,len(lst),chunk_size):
        yield lst[i:i+chunk_size]
        
x=data_loader(5,lst)
#print(type(x))
#print(x)
print(next(x))
print(next(x))
```

Map function in Python

```python
lst=[1,2,3,4,5,6]
def square(x):
    return x**2

lst=list(map(square,lst))
print(lst)
```

```python
#string
string="hello i am mahfuza"
string1=list(map(str.upper,string))
print(string1)
string1=" ".join(string1)
print(string1)
```

```python
#map with lambda
lst=[1,2,3,4,5]
lst=list(map(lambda x:x**2 if x%2==0 else x,lst))
print(lst)
```

Filter Function in Python
filter(2 parameter)
one boolean second iterable

```python
numbers=[x for x in range(100)]
#print(numbers)

even=list(filter(lambda x:x%2==0,numbers))
print(even)

#fifty_upper
fifty_upper=list(filter(lambda x:x>50,numbers))
print(fifty_upper)
```

```python
data=[0,1,'',None,"Hello",[],[1,2],True,False]

cleaned_data=list(filter(None,data))
print(cleaned_data)
```

```python
string="hello i am mahfuzaaa"
string1=list(map(str.upper,string))
string1=" ".join(string1)
print(string1)
vowel=list(filter(lambda x:x in'AEIOU',string1))
print(vowel)
```

```python
fruits=['aaa','bb','cccc','dddd']

filtering=list(filter(lambda x:len(x)>2,fruits))
print(filtering)
```

```python
from functools import reduce
lst=[1,2,3,4,5]
sum=reduce(lambda x,y:x+y,lst)
print(sum)
```
