String Splitting Joining and Formatting

```python
promt="What is your name?"

tokens=promt.split()
print(tokens)
print(type(tokens))

sentence="-".join(tokens)

print(sentence)
print(type(sentence))
```

list comprehension

```python
even=[]
for i in range(10):
    if i%2==0:
        even.append(i)
print(even)        
```

```python
even = [i for i in range(0,100,1) if i % 2 != 0]
print(even)
```
